## 3. Preprocessing Teks (Pembersihan Data)
Proses ini membersihkan teks komentar agar siap diolah oleh model Naive Bayes:
- **Case Folding:** Mengubah semua huruf menjadi huruf kecil.
- **Cleaning:** Menghilangkan tanda baca, angka, simbol, dan spasi ganda.
- **Stopword Removal:** Menghapus kata umum sambung bahasa Indonesia yang tidak bermakna sentimen.
- **Tokenisasi:** Memecah kalimat menjadi kumpulan kata (tokens).

```python
# Daftar kata umum (Stopwords) Bahasa Indonesia
stopwords_id = {
    'dan', 'yang', 'di', 'ini', 'itu', 'dengan', 'untuk', 'pada', 'ke', 'dari',
    'sama', 'saya', 'saat', 'ada', 'juga', 'adalah', 'karena', 'oleh', 'namun',
    'atau', 'sudah', 'bisa', 'akan', 'kita', 'dia', 'mereka', 'sebuah', 'telah',
    'ketika', 'agak', 'sekali', 'sangat', 'secara', 'harus', 'lagi', 'pun',
    'nya', 'bila', 'jika', 'hanya', 'cukup', 'kami', 'mengenai', 'bikin', 'jadi'
}

def preprocessing_teks(text):
    # 1. Case Folding
    text = str(text).lower()
    # 2. Cleaning Simbol & Tanda Baca
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    # 3. Stopword Removal
    tokens = text.split()
    filtered_tokens = [w for w in tokens if w not in stopwords_id and len(w) > 1]
    return ' '.join(filtered_tokens)

# Terapkan preprocessing ke DataFrame
df['komentar_bersih'] = df['komentar'].apply(preprocessing_teks)
df['tokens'] = df['komentar_bersih'].apply(lambda x: x.split())

print("✅ Preprocessing data berhasil diselesaikan!")
```

## 5. Pemodelan Naive Bayes & Evaluasi Model
- **Data Input:** `komentar_bersih`
- **Data Label:** `sentimen` (`marah` / `sedih`)
- **Split Data:** 160 Training (80%) dan 40 Testing (20%)
- **Algoritma:** Multinomial Naive Bayes (MultinomialNB)

```python
# Data Input dan Label
X = df['komentar_bersih']
y = df['sentimen']

# Split Train & Test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=40, random_state=42, stratify=y
)

print(f"Jumlah data training: {len(X_train)}")
print(f"Jumlah data testing: {len(X_test)}")

# Vektorisasi Bag of Words (CountVectorizer)
vectorizer = CountVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

print("\nBentuk data Training:", X_train_vec.shape)
print("Bentuk data Testing:", X_test_vec.shape)
print(f"Total Vocabulary Terbentuk: {len(vectorizer.vocabulary_)} kata")

# Menampilkan sampel Vocabulary seperti di video tugas Colab
sample_vocab = dict(list(vectorizer.vocabulary_.items())[:12])
print("\nVocabulary:")
print(sample_vocab)
```

```python
# Melatih Model Multinomial Naive Bayes
model_nb = MultinomialNB()
model_nb.fit(X_train_vec, y_train)

print("Model berhasil dilatih!")

# Prediksi Data Uji
y_pred = model_nb.predict(X_test_vec)

# Menghitung Akurasi
akurasi = accuracy_score(y_test, y_pred)
print(f"\nAkurasi Model: {akurasi:.4f} ({akurasi*100:.2f}%)")

# Classification Report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Visualisasi Confusion Matrix
cm = confusion_matrix(y_test, y_pred, labels=['marah', 'sedih'])

plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Marah', 'Sedih'], yticklabels=['Marah', 'Sedih'])
plt.xlabel('Prediksi Model', fontweight='bold')
plt.ylabel('Label Aktual (Sebenarnya)', fontweight='bold')
plt.title('Confusion Matrix Naive Bayes', fontweight='bold', pad=10)
plt.show()
```

## 6. Chatbot Sentimen (Input Komentar) & Tanggapan / Respon Otomatis
Fitur untuk memasukkan komentar baru, memprediksi sentimen (**marah** / **sedih**), dan chatbot secara otomatis memberikan tanggapan / respon yang tepat.

```python
def chatbot_karhutla(komentar_pengguna):
    # Preprocessing input pengguna
    clean_input = preprocessing_teks(komentar_pengguna)
    
    # Prediksi menggunakan model Naive Bayes
    vec_input = vectorizer.transform([clean_input])
    pred = model_nb.predict(vec_input)[0]
    proba = model_nb.predict_proba(vec_input)[0]
    confidence = max(proba) * 100
    
    # Generator Tanggapan dan Respon Chatbot Sesuai Sentimen
    if pred == 'marah':
        respon_chatbot = (
            "Kami memahami dan mendukung kemarahan serta kepedulian Anda. "
            "Pembiaran kebakaran hutan adalah kejahatan lingkungan serius! "
            "Aspirasi masyarakat seperti ini sangat penting untuk mendesak aparat penegak hukum "
            "(Gakkum KLHK & Kepolisian) agar segera menjatuhkan sanksi pidana dan mencabut izin perusahaan pembakar lahan! 🔴✊"
        )
    else:
        respon_chatbot = (
            "Hati kami turut pilu membaca kesedihan ini. Bencana kabut asap membawa dampak berat "
            "bagi anak-anak sekolah, lansia, dan ekosistem hutan kita. "
            "Prioritas mendesak saat ini adalah pendistribusian masker N95, pembukaan posko oksigen darurat gratis, "
            "serta pemulihan lahan terdampak. Semoga situasi segera pulih kembali. 😢🙏"
        )
        
    print("=" * 70)
    print("👤 Komentar Pengguna:", komentar_pengguna)
    print(f"🎯 Hasil Prediksi: {pred.upper()} (Tingkat Keyakinan: {confidence:.2f}%)")
    print("\n🤖 Tanggapan & Respon Chatbot:")
    print(respon_chatbot)
    print("=" * 70)

# Uji Coba Otomatis
print("=== CONTOH UJI COBA OTOMATIS ===")
chatbot_karhutla("Perusahaan pembakar lahan harus dihukum seberat-beratnya, jangan ada pembiaran!")
print()
chatbot_karhutla("Sedih sekali melihat kabut asap tebal, anak-anak sekolah sesak napas dan hewan liar mati.")
```
